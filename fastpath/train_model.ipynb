# Training the model

This is where most projects stop: a model that works, in a notebook, on one laptop.

Run every cell, then open `app.py` to see how it gets online.


## 1. The rule we want the model to learn

In a real project this would be labels from measurements. Here we generate them,
so the notebook runs anywhere with no data file.


```python
import random
import joblib
from sklearn.tree import DecisionTreeClassifier


def label(value, low, high):
    if low <= value <= high:
        return "normal"
    margin = (high - low) / 2 or 1
    if value < low - margin or value > high + margin:
        return "critical"
    return "warning"
```

## 2. One useful feature instead of three raw numbers

`position` is where the reading sits in its safe range: 0.0 at the bottom,
1.0 at the top, negative below it, above 1.0 over it.

This lets the model treat a body temperature and a motor temperature the same way.


```python
def features(value, low, high):
    span = (high - low) or 1
    return [(value - low) / span]
```

## 3. Make the training data


```python
random.seed(0)
X, y = [], []

for _ in range(3000):
    low   = random.uniform(0, 500)
    high  = low + random.uniform(1, 500)
    value = random.uniform(low - (high - low), high + (high - low))
    X.append(features(value, low, high))
    y.append(label(value, low, high))

len(X)
```

## 4. Train it


```python
model = DecisionTreeClassifier(max_depth=6).fit(X[:2500], y[:2500])
model
```

## 5. Check it on data it has never seen

Always score on data the model was not trained on. Scoring on the training
data tells you nothing.


```python
print(f"accuracy on unseen data: {model.score(X[2500:], y[2500:]):.1%}")
```

## 6. Save it

`model.pkl` **is** the model. It is the only thing that leaves this notebook.

Everything above ran once. From now on, `app.py` just loads this file.


```python
joblib.dump(model, "model.pkl")
```
